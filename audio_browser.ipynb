# Audio Browser

One cell per folder. Each cell lists every audio file in that folder with an inline player.

**How to use**
1. Edit `AUDIO_ROOTS` in the setup cell — add every folder whose outputs you want to browse.
2. Run the setup cell once.
3. Run any folder cell to see + play its audio. Re-run after new runs finish to refresh.

Supported: `.wav .mp3 .flac .ogg .m4a .opus .aac`

```python
# --- Setup: run once ---
from pathlib import Path
from IPython.display import Audio, display, HTML, Markdown

# EDIT THIS: list every folder that holds run outputs you want to browse.
AUDIO_ROOTS = [
    Path('outputs'),
    Path('runs'),
    Path('samples'),
]

AUDIO_EXT = {'.wav', '.mp3', '.flac', '.ogg', '.m4a', '.opus', '.aac'}

def show_folder(folder, recursive=False):
    folder = Path(folder)
    if not folder.exists():
        display(Markdown(f"_Folder not found: `{folder}`_"))
        return
    it = folder.rglob('*') if recursive else folder.iterdir()
    files = sorted(p for p in it if p.is_file() and p.suffix.lower() in AUDIO_EXT)
    display(Markdown(f"### `{folder}`  —  {len(files)} file(s)"))
    if not files:
        display(Markdown('_no audio_'))
        return
    for f in files:
        display(Markdown(f"**{f.name}**  · `{f.relative_to(folder)}`  · {f.stat().st_size/1024:.1f} KB"))
        display(Audio(str(f)))

print('Ready. Run any folder cell below.')
```

## outputs/

```python
show_folder('outputs', recursive=True)
```

## runs/

```python
show_folder('runs', recursive=True)
```

## samples/

```python
show_folder('data/listen', recursive=True)
```

## Auto-discover: one section per subfolder

Useful when a run writes many subfolders you don't want to list by hand.

```python
ROOT = Path('outputs')  # change to whichever run root
if ROOT.exists():
    subs = sorted([p for p in ROOT.iterdir() if p.is_dir()])
    for s in subs:
        show_folder(s, recursive=True)
else:
    print(f'{ROOT} does not exist yet')
```
